```python
from transformers import pipeline 
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
docs = [
    'I love Ai',
    'The wheather is amazing',
    'Machine learning is amazing'
]
embeddings = model.encode(docs)
print(embeddings.shape)
print(embeddings)
```

```python
query = 'Artificial Intelligence'

query_embedding = model.encode(query)
print(query_embedding.shape)
print(query_embedding)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
# apply cosine similarity with each document to the query
cosine_similarities = cosine_similarity([query_embedding], embeddings)
print(cosine_similarities)

# get the index of the most similar document
most_similar_doc_index = np.argmax(cosine_similarities)
print(most_similar_doc_index)
print(docs[most_similar_doc_index])
```
